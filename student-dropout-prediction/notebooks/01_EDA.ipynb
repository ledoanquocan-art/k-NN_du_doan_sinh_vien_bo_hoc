```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc file dữ liệu thô (để hiển thị nhãn chữ cho dễ nhìn)
df = pd.read_csv('../data/raw/data.csv', sep=';')

# Biểu đồ 1: Phân bổ 3 loại nhãn Target
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Target', palette='Set2')
plt.title('Phân bố tình trạng sinh viên (Tổng: 4424 mẫu)')
plt.ylabel('Số lượng')
plt.show()

# Biểu đồ 2: Sinh viên còn nợ học phí có dễ bỏ học không?
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Tuition fees up to date', hue='Target', palette='Set1')
plt.title('Tác động của việc Đóng học phí đúng hạn đến việc Bỏ học')
plt.xlabel('Đóng học phí đúng hạn (0: Không, 1: Có)')
plt.ylabel('Số lượng')
plt.show()
```
